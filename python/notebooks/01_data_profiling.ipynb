### A. Check whether order-item identifiers are unique

### C. Check order and shipping date logic

```python
order_date = pd.to_datetime(
    df["order date (DateOrders)"], errors="coerce"
)

shipping_date = pd.to_datetime(
    df["shipping date (DateOrders)"], errors="coerce"
)

date_check = pd.DataFrame({
    "order_date": order_date,
    "shipping_date": shipping_date
})

date_check["shipping_before_order"] = (
    date_check["shipping_date"] < date_check["order_date"]
)

print(
    "Shipping dates before order dates:",
    date_check["shipping_before_order"].sum()
)

date_check["actual_days_between"] = (
    date_check["shipping_date"] - date_check["order_date"]
).dt.total_seconds() / 86400

display(date_check["actual_days_between"].describe())
```

# Cleaning Script

```python
from pathlib import Path
import pandas as pd

# Check the notebook's current working directory
print("Current working directory:", Path.cwd())

# Set the project root manually based on your folder structure
PROJECT_ROOT = Path.cwd()

# If your notebook runs from python/notebooks, use:
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent.parent

RAW_FILE = (
    PROJECT_ROOT
    / "data"
    / "raw"
    / "DataCoSupplyChainDataset.csv"
)

PROCESSED_DIR = PROJECT_ROOT / "data" / "processed"
OUTPUT_FILE = PROCESSED_DIR / "dataco_cleaned.csv"

print("Project root:", PROJECT_ROOT)
print("Raw file exists:", RAW_FILE.exists())
print("Raw file path:", RAW_FILE)
```

```python
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

df = pd.read_csv(
    RAW_FILE,
    encoding="latin-1",
    low_memory=False
)
```

### Step 1: Validate the order-item key

```python
key_col = "Order Item Id"

assert df[key_col].notna().all(), "Missing order-item IDs detected."
assert df[key_col].is_unique, "Duplicate order-item IDs detected."

print("Order Item Id validation passed!")
print("Total rows:", len(df))
print("Unique order-item IDs:", df[key_col].nunique())
```

### Step 2: Remove unsuitable columns

```python
initial_rows = len(df)
initial_columns = len(df.columns)

columns_to_drop = [
    "Product Description",
    "Order Zipcode",
    "Customer Email",
    "Customer Fname",
    "Customer Lname",
    "Customer Street",
    "Customer Password",
]

columns_to_drop = [
    col for col in columns_to_drop
    if col in df.columns
]

# Drop the identified columns
df = df.drop(columns=columns_to_drop)

# Drop Product Status only if it is constant
if "Product Status" in df.columns:
    if df["Product Status"].nunique(dropna=False) == 1:
        df = df.drop(columns=["Product Status"])

print("Initial rows:", initial_rows)
print("Initial columns:", initial_columns)
print("Columns removed:", columns_to_drop)
print("Processed columns:", len(df.columns))
print("Rows retained:", len(df))
```

### Step 3: Convert the date columns

```python
date_columns = [
    "order date (DateOrders)",
    "shipping date (DateOrders)",
]

for col in date_columns:
    if col in df.columns:
        df[col] = pd.to_datetime(df[col], errors="raise")

print(df[date_columns].dtypes)
```

```python
invalid_dates = (
    df["shipping date (DateOrders)"]
    < df["order date (DateOrders)"]
)

print("Shipping dates before order dates:", invalid_dates.sum())

assert not invalid_dates.any(), "Shipping-before-order records detected."

print("Date validation passed!")
```

### Step 4: Validate and save the cleaned dataset

```python
# Confirm no rows were accidentally removed
assert len(df) == initial_rows, "Unexpected row-count change."

# Revalidate the key after cleaning
assert df["Order Item Id"].is_unique
assert df["Order Item Id"].notna().all()

# Save to a separate processed file
df.to_csv(OUTPUT_FILE, index=False)

print("\nCleaning completed successfully!")
print("Processed rows:", len(df))
print("Processed columns:", len(df.columns))
print("Output file:", OUTPUT_FILE)
print("File exists:", OUTPUT_FILE.exists())
```

```python
# Reload the saved file independently
cleaned_df = pd.read_csv(
    OUTPUT_FILE,
    encoding="latin-1",
    low_memory=False
)

print("Saved file rows:", cleaned_df.shape[0])
print("Saved file columns:", cleaned_df.shape[1])
print(
    "Missing Order Item Id values:",
    cleaned_df["Order Item Id"].isna().sum()
)
print(
    "Duplicate Order Item Id values:",
    cleaned_df["Order Item Id"].duplicated().sum()
)
print(
    "Exact duplicate rows:",
    cleaned_df.duplicated().sum()
)

print("\nRemaining missing values:")
missing_after_cleaning = (
    cleaned_df.isna()
    .sum()
    .sort_values(ascending=False)
)

display(missing_after_cleaning[missing_after_cleaning > 0])
```
